```python
#demo: Convert ATCG to Vectors and Understand Positional Importance
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Step1: Define DNA Sequence
sequence = "ATCGATCG"
print("="*60)
print("Original DNA Sequence")
print("="*60)
print(sequence)
```

```python
#Step2 :One-Hot encoding Dictionary
```

```python
encoding ={
    "A":[1,0,0,0],
    "T":[0,1,0,0],
    "C":[0,0,1,0],
    "G":[0,0,0,1]
}
```

```python
#Step3: Convert DNA Sequence into Vectors
encoded_sequence = np.array([encoding[base] for base in sequence])
print("\n"+"="*60)
print("One-Hot Encoded DNA Sequence")
print("="*60)
print(encoded_sequence)
```

```python
#Step 4: Create DataFrame
df = pd.DataFrame(encoded_sequence, columns=["A","T","C","G"])
print("\n"+"="*60)
print("DataFrame")
print("="*60)
print(df)
```

```python
#Step 5: Show Positional Information
position_df = pd.DataFrame({
    "Position": range(1,len(sequence)+1),
    "Base":list(sequence)
})
print("\n"+"="*60)
print("Positional Information")
print("="*60)
print(position_df)
```

```python
#Start 6: Visualize Encoded Matrix
plt.figure(figsize=(8,4))
plt.imshow(encoded_sequence,aspect='auto')
plt.colorbar()
plt.title("One-Hot Encoded DNA Sequence")
plt.xlabel("Nucleotides")
plt.ylabel("Sequence Position")
plt.xticks([0,1,2,3],["A","T","C","G"])
plt.show()
```

```python
#Step 7: Mutation Example
mutated_seuqence ="ATCAATCG"
print("\n"+"="*60)
print("Mutation Analysis")
print("="*60)
print(f"Original Sequence: {sequence}")
print(f"Mutated Sequence: {mutated_seuqence}")
for i in range(len(sequence)):
    if sequence[i] != mutated_seuqence[i]:
        print(f"\nMutation found at position {i+1}")
        print(f"{sequence[i]}->{mutated_seuqence[i]}")
```

```python
#Step 8: Encode Mutated Sequence
mutated_encoded = np.array([encoding[base] for base in mutated_seuqence])
print("\n"+"="*60)
print("One-Hot Encoded Mutated Sequence")
print("="*60)
print(mutated_encoded)
```

```python
#Step9: Compare Original and Mutated Vectors
print("\n"+"="*60)
print("Difference between Original and Mutated Vectors")
print("="*60)
difference = encoded_sequence - mutated_encoded
print(difference)
```

```python
#Step 10: Prepare Deep Learning Input
x = np.expand_dims(encoded_sequence,axis=0)
print("\n"+"="*60)
print("Deep Learning Input Shape")
print("="*60)
print(x.shape)
print("\n Interpretation:")
print("1 Sequence")
print(f"{x.shape[1]} nucleotides")
print(f"{x.shape[2]} features(A,T,G,C)")
```

```python
#Step 11: Display Tensor Format
print("\n"+"="*60)
print("Deep Learning Input Tensor Format")
print("="*60)
print(x)
print("\n Demo Completed Successfully!")
```
