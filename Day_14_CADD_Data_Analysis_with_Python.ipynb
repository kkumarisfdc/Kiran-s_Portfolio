```python
from rdkit import Chem
from rdkit.Chem import Descriptors

smiles_list = {"Compound CID: 9549299":"C1CC1C(=O)NC2=CC=CC(=C2)NC3=NC=NC(=C3)NC4=CC=CC(=C4)C(F)(F)F",
               "Compound CID: 117918742":"C[C@@H](CN2C1=C(C3=C(N=CN=C32)N)C4=CC5=CC=CC=C5N=C4)NC(=O)C=C",
               "Compound CID: 9843206":" COC1=C(C=C2C(=C1)C(=NC=N2)NC3=CC=C(C=C3)OCC4=CC=CC=C4)OC",
               "Compound CID: 121280087":"CN(C)CCN(C)C1=CC(=C(C=C1NC(=O)C=C)NC2=NC=CC(=N2)C3=CN(C4=CC=CC=C43)C5CC5)OC"
               }
for name,smi in smiles_list.items():
  mol  = Chem.MolFromSmiles(smi)
  if mol is not None:
    print(Descriptors.ExactMolWt(mol))

    mw  =  Descriptors.MolWt(mol)
    hbd = Descriptors.NumHDonors(mol)
    hba = Descriptors.NumHAcceptors(mol)
    tpsa = Descriptors.TPSA(mol)
    rings = Descriptors.RingCount(mol)

    print(f"Compound Name: {name}")
    print(f"SMILES: {smi}")
    print(f"Molecular Weight: {mw}")
    print(f"Number of Hydrogen Bond Donors: {hbd}")
    print(f"Number of Hydrogen Bond Acceptors: {hba}")
    print(f"Topological Polar Surface Area: {tpsa}")
    print(f"Number of Rings: {rings}")
    print()
  else:
    print(f"Error: Could not parse SMILES for {name}: {smi}")
    print()
```
