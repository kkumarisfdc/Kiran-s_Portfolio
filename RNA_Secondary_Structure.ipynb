```python
import RNA
```

```python
gRNA = "GAGUCCGAGCAGAAGAAGAAGGGGGGGGGGGGGGGGGGGG"

print("Guide RNA sequence:")
print(gRNA)
```

```python
structure,mfe = RNA.fold(gRNA)

print("Predicted structure:")
print(structure)
print("Minimum Free Energy:")
print(mfe)
```

```python
pairs  = structure.count("(")
print("Number of Base Pairs:",pairs)
```

```python
print("Delta G: =",mfe,"kcal/mol")
if(mfe<-10):
  print("Highly Stable Structure")
elif mfe<-5:
  print("Stable Structure")
else:
  print("Moderately Stable Structure")
```

```python
guides = [

"GGGAAAUUUCCC",

"GCGCGCAAAAUUU",

"GAUCGGAUCGGAUCG",

"GAGUCCGAGCAGAAG"
]
for seq in guides:
  structure,mfe  = RNA.fold(seq)
  print("Sequences:")
  print(seq)
  print("Structures:")
  print(structure)
  print("Minimum Free Energy(delta G):")
  print(mfe)
  print(".......................................")
```

```python
structures =[]
energies =[]
for seq in guides:
  ss,mfe = RNA.fold(seq)
  structures.append(ss)
  energies.append(mfe)
```

```python
for i in range(len(guides)):
  print("Guide",i+1)
  print("Sequence:",guides[i])
  print("Structures:",structures[i])
  print("Minimum Free Energy(delta G):",energies[i])
  print("................................................")
```

```python
import matplotlib.pyplot as plt
```

```python
labels =["g1","g2","g3","g4"]
plt.figure(figsize=(7,4))
plt.bar(labels,energies)
plt.xlabel("Guide RNA")
plt.ylabel("delta G (kcal/mol)")
plt.title("RNA Folding Stability")
plt.show()
```

```python
fc = RNA.fold_compound(gRNA)
structure, mfe = fc.mfe()
print(structure)
print(mfe)
```

```python
import pandas as pd
df = pd.DataFrame({
    'Sequence:':guides,
    'Structure':structures,
    'delta G':energies
})
df
```
