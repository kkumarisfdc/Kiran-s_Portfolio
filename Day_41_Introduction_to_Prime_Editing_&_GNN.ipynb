```python
# Sample DNA sequence
dna ="ATGGCTAGCGGATCGATCGGATCCGATCG"
print("Original DNA Sequence")
print(dna)
```

```python
print("Searching for PAM(NGG)....\n")
for i in range(len(dna)-2):
  if dna[i+1:i+3]=="GG":
    print(f"PAM found at a postions {i}: {dna[i:i+3]}")
```

```python
positon =12
edited = dna[:positon] + "T" +dna[positon+1:]
print("Original :",dna)
print("Edited :",edited)
```

```python
spacer = dna[2:22]
pbs = dna[22:30]
rtt= edited[10:28]
print("Spacer: ",spacer)
print("PBS :",pbs)
print("RTT :",rtt)
```

```python
import pandas as pd
data ={
    "GuideRNA":
    ["Guide1","Guide2","Guide3","Guide4"],
    "GC_Content":
    [40,55,65,48],
    "Efficiency":
    [72,91,80,86]
}

df= pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.bar(df["GuideRNA"],df["Efficiency"])
plt.title("Graph-CRISPR Prediction")
plt.xlabel("Guide RNA")
plt.ylabel("Predicted Efficiency (%)")
plt.show()
```

```python
import networkx as nx
G= nx.path_graph(10)
labels = {i:dna[i] for i in range(10)}
plt.figure(figsize=(8,2))
nx.draw(G,
        labels=labels,
        with_labels=True,
        node_color = "lightgreen"
        )
```

```python
best = df.loc[df["Efficiency"].idxmax()]
print("Best Guide DNA")
print(best)
```
