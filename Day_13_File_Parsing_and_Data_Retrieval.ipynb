```python
import requests
from Bio import SeqIO
from io import StringIO

#ENA accession number for SARS -CoV -2
ena_accession = "LR757998"
#Construct DBFetch URL
url = f"https://www.ebi.ac.uk/Tools/dbfetch/dbfetch?db=embl&id={ena_accession}&format=fasta&style=raw"
#fetch data
response = requests.get(url)

if response.ok:
  fasta_text =  response.text
  print("First 500 characters:\n",fasta_text[:500])
  #parse with Biopython
  record = SeqIO.read(StringIO(fasta_text),"fasta-pearson")
  print("\n Record ID:",record.id)
  print("\n Sequence Length:",len(record.seq))
else:
  print("Failed to Fetch sequence, status code",response.status_code)
```

```python
#Reading and writing fasta files
#import required classes
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
```

```python


# Define example sequences
seq1 = Seq("ATGGCTTAGCTAGCTACGATCGTAACTGATC")
seq2 = Seq("ATGCGATCGATCGATCGATCGGCTAGCTAGCTAGCTA")
seq3 = Seq("ATGCCCGGGAAATTTCCCGGGAAATTTCCCGGG")

# Create SeqRecord objects with metadata
record1 = SeqRecord(seq1, id="GeneA", description="Synthetic gene A sequence")
record2 = SeqRecord(seq2, id="GeneB", description="Synthetic gene B sequence")
record3 = SeqRecord(seq3, id="GeneC", description="Synthetic gene C sequence")

#Display record details
for r in [record1, record2, record3]:
  print(f"Record ID:{r.id}")
  print(f"Description:{r.description}")
  print(f"Sequence: {r.seq} \n")


```

```python
#write the first record to a single FASTA file
SeqIO.write(record1, "single_geneA.fasta","fasta")

#Show the saved file
!head single_geneA.fasta
```

```python
#writing multiple sequences (Multi-FASTA file)
records =[record1,record2,record3]
#write all sequences into one fasta file
SeqIO.write(records,"multi_genes.fasta","fasta")

!head multi_genes.fasta
```

```python
#Append a New Sequence to Existing FASTA
# Create a new sequence
seq4 = Seq("ATGTTTTTGGGCCCAAAATTTGGGCCC")
record4 = SeqRecord(seq4, id="GeneD", description="Synthetic gene D sequence")

# Read old records and add new one
existing_records = list(SeqIO.parse("multi_genes.fasta", "fasta"))
existing_records.append(record4)

# Write back to same file (overwrites with all records)
SeqIO.write(existing_records, "multi_genes.fasta", "fasta")

!head multi_genes.fasta
```

```python
#Parsing Genbank files
from Bio import SeqIO, Entrez

Entrez.email = "your_email@example.com"

# Download GenBank record directly from NCBI
accession = "NC_045512.2"
handle = Entrez.efetch(db="nucleotide", id=accession, rettype="gb", retmode="text")

# Parse the GenBank record using SeqIO
genome_record = SeqIO.read(handle, "genbank")
handle.close()

print(f"Successfully fetched {accession} from NCBI!")
print(f"Record ID: {genome_record.id}")
print(f"Description:{genome_record.description}")
```

```python
#Define a Function to Search for a CDS by Qualifier
def get_cds_feature_with_qualifier_value(seq_record, name, value):

  """Find a CDS feature in the sequence record with a specific qualifier value.

Args:
seq_record: SeqRecord object (parsed from GenBank)
name: Qualifier name (e.g., 'old_locus_tag', 'locus_tag', 'gene')
value: Qualifier value to match

Returns:
The matching feature (SeqFeature) or None if not found.
"""
  for feature in seq_record.features:
    if feature.type == "CDS" and value in feature.qualifiers.get(name, []):
      return feature
  return None
```

```python
qualifier_name = "locus_tag"
qualifier_value = "GU280_gp01"

cds_feature = get_cds_feature_with_qualifier_value(genome_record, qualifier_name, qualifier_value)

if cds_feature:
  print("Found CDS feature!")
  print("Gene:", cds_feature.qualifiers.get("gene", ["N/A"])[0])
  print("Product:", cds_feature.qualifiers.get("product", ["N/A"])[0])
  print("Location:", cds_feature.location)
  print("Length (bp):", len(cds_feature.extract(genome_record.seq)))
else:
  print("No CDS found with the specified locus_tag")
```

The `locus_tag` 'GU280_gp01' is not present in the `genome_record`. Let's inspect the available `locus_tag` values in the `genome_record` to find a valid one.

```python
# Collect and print all available locus_tags
all_locus_tags = []
for feature in genome_record.features:
    if feature.type == "CDS" and "locus_tag" in feature.qualifiers:
        all_locus_tags.extend(feature.qualifiers["locus_tag"])

print("Available Locus Tags:")
for tag in sorted(list(set(all_locus_tags))):
    print(f"- {tag}")

if not all_locus_tags:
    print("No CDS features with 'locus_tag' qualifier found in the record.")
```

Now that we have a list of available `locus_tag` values, let's pick one (e.g., 'GU280_gp02') and try to find its corresponding CDS feature.

```python
qualifier_name = "locus_tag"
# Update with a known locus_tag from the list above, for example 'GU280_gp02'
qualifier_value = "GU280_gp02"

cds_feature_updated = get_cds_feature_with_qualifier_value(genome_record, qualifier_name, qualifier_value)

if cds_feature_updated:
  print("Found CDS feature with updated locus_tag!")
  print("Gene:", cds_feature_updated.qualifiers.get("gene", ["N/A"])[0])
  print("Product:", cds_feature_updated.qualifiers.get("product", ["N/A"])[0])
  print("Location:", cds_feature_updated.location)
  print("Length (bp):", len(cds_feature_updated.extract(genome_record.seq)))
else:
  print("No CDS found with the specified locus_tag (even after update). Please check the available locus tags.")
```
