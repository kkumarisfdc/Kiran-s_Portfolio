```python
#import: align function from biopython package
from Bio import Align
```

```python
Aligner  = Align.PairwiseAligner(match_score=1.0)
```

```python
#Defining to sequences for alignment
SeqA = "GAATCCAG"
SeqB = "AACAG"
#Call the aligner to assign score
score = Aligner.score(SeqA,SeqB)
```

```python
#To print the alignment score
print(f"The alignment score is {score}")
```

```python
#For the exact alignment map in global alignment
Alignments  = Aligner.align(SeqA, SeqB)
#Using for loop
for alignment in Alignments:
    print(alignment)
```

```python
#For the alignment score in local alignment
Align.mode ="local"
Seq1= "ATTGCGATTAAA"
Seq2 = "GCG"
```

```python


Local_score = Aligner.score(Seq1,Seq2)
print(f"The local alignment score is {Local_score}")
```

```python

#For the alignment score in local alignment
Align_local  = Aligner.align(Seq1, Seq2)
for LA in Align_local:
    print(LA)
```

```python
from Bio import SeqIO
sequences =[]
for seq_record in SeqIO.parse('ls_orchid.fasta','fasta'):
  sequences.append(seq_record.seq)
sequences
```

```python
#History of sequence lengths
sizes = [len(rec) for rec in sequences]
import matplotlib.pyplot as plt
plt.hist(sizes,bins=20)
plt.xlabel("Sequence Length (bp)")
plt.ylabel("Count")
plt.title(f"{len(sizes)} Orchid Sequences\n Lengths{min(sizes)} to {max(sizes)}")
plt.show()
```
