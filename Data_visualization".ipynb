```python
# -*- coding: utf-8 -*-

"""colab-"Data visualization"-dr_bhupender-23_06_2026.ipynb

Automatically generated by Colab.

Step 1: Loading our Dataset

Visualizing Raw CRISPR Screen Matrices

In this tutorial, we will work with CRISPR-Cas9 dependency screening matrix.

When you download a matrix from portals like CCLE, it arrives as a massive table: **Rows represent Patient Cell Lines, and Columns represent Human Genes**. The values represent editing knockout effects (scores below 0 imply that knocking out the gene compromises cell survival).

"""

# Import foundational data science and plotting frameworks

import matplotlib.pyplot as plt

import pandas as pd

import seaborn as sns
```

```python
# Establish a uniform, clean style for scientific publishing

sns.set_theme(style="ticks")

# Load a genuine public CRISPR screen matrix (Cell Lines x Genes)

data_url = "https://raw.githubusercontent.com/kimmo1019/DeepCDR/master/data/CCLE/genomic_expression_561celllines_697genes_demap_features.csv"

raw_matrix = pd.read_csv(data_url, index_col=0)

# Display a preview of the raw, untouched biological matrix

print(f"Matrix Scale: {raw_matrix.shape[0]} Cell Lines by {raw_matrix.shape[1]} Genes")

raw_matrix.head()

"""Step 2 - Evaluating Global Matrix Value Spread (Histogram)"""

# 1. Establish the visualization canvas dimension properties

plt.figure(figsize=(8, 5))
```

```python
# 2. Plot the raw distribution of the 'MYC' column across all rows

sns.histplot(

data=raw_matrix,

x="MYC", # Name of the target gene column from our loaded matrix

bins=30, # Group continuous data points into 30 uniform bars

kde=True, # Plot a smooth probability density trendline

color="#2b5c8f", # Apply a professional slate-blue color

)
```

```python
# 3. Layer precise axis variables and contextual headings

plt.title(

"Global Screening Variance Profile for Human MYC Gene",

fontsize=13,

fontweight="bold",

)

plt.xlabel("mRNA Expression Amplitude", fontsize=11)

plt.ylabel("Patient Cell Line Count (Frequency)", fontsize=11)
```

```python


# Establish a uniform, clean style for scientific publishing

sns.set_theme(style="ticks")

# Load a genuine public CRISPR screen matrix (Cell Lines x Genes)

data_url = "https://raw.githubusercontent.com/kimmo1019/DeepCDR/master/data/CCLE/genomic_expression_561celllines_697genes_demap_features.csv"

raw_matrix = pd.read_csv(data_url, index_col=0)

# Display a preview of the raw, untouched biological matrix

print(f"Matrix Scale: {raw_matrix.shape[0]} Cell Lines by {raw_matrix.shape[1]} Genes")

raw_matrix.head()

"""Step 2 - Evaluating Global Matrix Value Spread (Histogram)"""

# 1. Establish the visualization canvas dimension properties

plt.figure(figsize=(8, 5))

# 2. Plot the raw distribution of the 'MYC' column across all rows

sns.histplot(

data=raw_matrix,

x="MYC", # Name of the target gene column from our loaded matrix

bins=30, # Group continuous data points into 30 uniform bars

kde=True, # Plot a smooth probability density trendline

color="#2b5c8f", # Apply a professional slate-blue color

)

# 3. Layer precise axis variables and contextual headings

plt.title(

"Global Screening Variance Profile for Human MYC Gene",

fontsize=13,

fontweight="bold",

)

plt.xlabel("mRNA Expression Amplitude", fontsize=11)

plt.ylabel("Patient Cell Line Count (Frequency)", fontsize=11)

# 4. Save both crisp PNG and structural vector PDF files to your workspace directory

plt.savefig("crispr_raw_histogram.png", dpi=300, bbox_inches="tight")

plt.savefig("crispr_raw_histogram.pdf", bbox_inches="tight")

# 5. Render the final plot onto the screen

plt.show()

"""Step 3 - Comparing Multiple Genes side-by-side (Boxplot)"""

# 1. Initialize canvas attributes for box tracking

plt.figure(figsize=(9, 6))

# 2. Select specific gene columns and pass the slice directly to the boxplot tool

# This plots each column side-by-side

sns.boxplot(

data=raw_matrix[["TP53", "RB1", "EGFR", "MDM2"]],

palette="Set2", # Implement publication-standard grouping colors

width=0.4, # Adjust geometry widths to preserve clean white spaces

)

# 3. Define layout text coordinates

plt.title(

"Side-by-Side Efficacy Comparison of Key Gene Profiles",

fontsize=13,

fontweight="bold",

)

plt.xlabel("Target Human Genes", fontsize=11)

plt.ylabel("Expression Distribution", fontsize=11)

# 4. Save high-resolution variants for your presentation or manuscript drafts

plt.savefig("crispr_raw_boxplot.png", dpi=300, bbox_inches="tight")

plt.savefig("crispr_raw_boxplot.pdf", bbox_inches="tight")

# 5. display the finished plot object

plt.show()

"""Step 4 - Displaying the Screen Grid Directly (Heatmap)"""

# 1. Grab a clean 15x15 row/column block directly from the raw downloaded matrix using .iloc

# This is a direct window into the raw dataset with no formatting changed

heatmap_subset = raw_matrix.iloc[0:15, 0:15]

# 2. Establish frame boundaries for spatial coordinate grid visualization

plt.figure(figsize=(11, 9))

# 3. Render the expression intensity matrix using a sequential heatmap color layout

sns.heatmap(

data=heatmap_subset, # Feed the raw matrix slice directly to the plotting engine

cmap="mako", # 'mako' is a professional sequential palette (Dark = High expression, Light = Low expression)

annot=True, # Embed the exact quantitative numbers inside every matrix block

fmt=".2f", # Limit numeric decimal points for visual neatness

linewidths=0.5, # Add clean structural channels between grid squares

cbar_kws={

"label": "Transcript Abundance Level"

}, # Label the scale bar

)

# 4. Add formal figure descriptions

plt.title(

"Raw Cross-Section Genomic Matrix Pattern Layout",

fontsize=13,

fontweight="bold",

)

plt.xlabel("Target Human Genes (Columns)", fontsize=11)

plt.ylabel("Human Cancer Cell Lines (Rows)", fontsize=11)

# 5. Save publication-ready graphics assets directly to your environment workspace

plt.savefig("crispr_raw_heatmap.png", dpi=300, bbox_inches="tight")

plt.savefig("crispr_raw_heatmap.pdf", bbox_inches="tight")

# 6. Render the final visual asset

plt.show()
```
