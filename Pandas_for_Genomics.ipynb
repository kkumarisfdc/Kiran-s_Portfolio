```python
import pandas as pd
```

```python
data = {
    "gRNA":["g1","g2","g3","g4","g5","g6","g7","g8","g9","g10"],
    "Gene":["TP53","TP53","BRCA1", "BRCA1", "EGFR","EGFR","MYC","MYC","KRAS","KRAS"],
    "Efficiency":[0.82,0.75,0.65,0.92,0.95,0.78,0.88,0.72,0.97,0.81]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
df.head()
```

```python
df["Gene"]
```

```python
high_eff = df[df["Efficiency"]>0.80]
high_eff
```

```python
threshold = df["Efficiency"].quantile(0.95)
print(threshold)
```

```python
top_guides = df[
    df["Efficiency"] >= threshold

]
top_guides
```

```python
df.sort_values(
    by="Efficiency",
    ascending=False
)
```

```python
gene_summary  = df.groupby(
    "Gene"
)["Efficiency"].mean()
gene_summary
```

```python
df.groupby(
    "Gene"
)["Efficiency"].agg(["mean","max","min"])
```

```python
gene_info = pd.DataFrame(
    {
        "Gene":["TP53","BRCA1","EGFR","MYC","KRAS"],
        "Chromosome":[17,17,17,8,12]
    }
)
gene_info
```

```python
merged = pd.merge(
    df,
    gene_info,
    on="Gene"
)

merged
```
