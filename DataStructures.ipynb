```python
#Simulate Random DNA and Compute GC Content Vectorized

#First, let's import the necessary libraries: numpy for random choice and array operations, and collections.Counter to easily count nucleotide occurrences.



import numpy as np

from collections import Counter
```

```python
# Function to simulate a random DNA sequence

def simulate_random_dna(length):

    """Generates a random DNA sequence of a given length."""

    nucleotides = ['A', 'T', 'C', 'G']

    return ''.join(np.random.choice(nucleotides, length))



# Function to compute GC content of a single DNA sequence

def compute_gc_content(dna_sequence):

    """Computes the GC content (percentage) of a DNA sequence."""

    if not dna_sequence:

        return 0.0

    counts = Counter(dna_sequence.upper())

    gc_count = counts['G'] + counts['C']

    total_bases = sum(counts.values())

    return (gc_count / total_bases) * 100



print("Helper functions `simulate_random_dna` and `compute_gc_content` defined.")
```

```python


num_sequences = 100

sequence_length = 500



#Now, let's simulate multiple DNA sequences and compute their GC content in a vectorized fashion using NumPy's array operations or a list comprehension, which is often efficient enough for Python lists.



# Simulate multiple DNA sequences

dna_sequences = [simulate_random_dna(sequence_length) for _ in range(num_sequences)]



# Compute GC content for all sequences in a vectorized way using list comprehension

gc_contents = [compute_gc_content(seq) for seq in dna_sequences]



# Display the first few sequences and their GC content

print(f"Simulated {num_sequences} DNA sequences of length {sequence_length}.")

print("\nFirst 5 sequences and their GC content:")

for i in range(5):

    print(f"Sequence {i+1}: {dna_sequences[i][:30]}... (GC Content: {gc_contents[i]:.2f}%) ")



# Optionally, you can convert to a NumPy array for further numerical analysis

gc_contents_np = np.array(gc_contents)

print(f"\nMean GC content: {np.mean(gc_contents_np):.2f}%")

print(f"Standard deviation of GC content: {np.std(gc_contents_np):.2f}%")

```
