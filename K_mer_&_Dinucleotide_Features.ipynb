```python
import pandas as pd
from collections import Counter

grna_data = {
"gRNA_ID": [
"gRNA1","gRNA2","gRNA3","gRNA4","gRNA5",
"gRNA6","gRNA7","gRNA8","gRNA9","gRNA10"

],

"Sequence":[
"GAGTCCGAGCAGAAGAAGAA",
"GCTGATCGTAGCTAGCTAGC",
"ATCGGCTAGCTAGGCTAGCT",
"CGTAGCTAGCGATCGATGCA",
"TCGATCGATCGTAGCTAGCA",
"AGCTAGCTAGCATCGATCGA",
"CGATCGTAGCTAGCTAGCTA",
"GCTAGCTAGCATCGTAGCTA",
"TAGCTAGCTAGCGATCGATC",
"ATCGTAGCTAGCATCGTAGC"
]
}

df = pd.DataFrame(grna_data)

df
```

```python
df = pd.concat([df]*10, ignore_index=True)

df["gRNA_ID"]=["gRNA_"+str(i+1) for i in range(len(df))]

df.head()
```

```python
def get_dinucleotides(sequence):

    dinucleotides = []

    for i in range(len(sequence)-1):

        dinucleotides.append(sequence[i:i+2])

    return dinucleotides
```

```python
sequence= "ATCGAT"

pairs = get_dinucleotides(sequence)

print(pairs)
```

```python
def dinucleotide_frequency(sequence):
    pairs = get_dinucleotides(sequence)

    counts = Counter(pairs)

    total = len(pairs)

    frequency = {}

    for pair in counts:

        frequency[pair] = counts [pair]/total

    return frequency
```

```python

sequence = "ATCGAT"

freq = dinucleotide_frequency(sequence)

print(freq)
```

```python

bases = ["A","T","G","C"]
all_pairs = []

for b1 in bases:
  for b2 in bases:
    all_pairs.append(b1+b2)
print(all_pairs)
```

```python
feature_table =[]
for seq in df["Sequence"]:
  freq = dinucleotide_frequency(seq)
  row =[]
  for pair in all_pairs:
    if pair in freq:
      row.append(freq.get(pair,0))
    else:
      row.append(0)
  feature_table.append(row)
```

```python
feature_df = pd.DataFrame(feature_table,columns=all_pairs)
feature_df.head()
```

```python
final_df = pd.concat([df,feature_df],axis=1)
final_df.head()
```

```python
def nucleotide_frequency(sequence):

  count = Counter(sequence)

  total = len(sequence)

  freq = {}

  for base in ["A","T","G","C"]:

    freq[base] = count[base]/total

  return freq
```

```python
def odds_ratio(sequence):

  mono = nucleotide_frequency(sequence)

  di = dinucleotide_frequency(sequence)

  odds = {}

  for pair in all_pairs:
    observed = di.get(pair,0)

    expected = mono[pair[0]]*mono[pair[1]]

    if expected == 0:

      odds [pair]=0

    else:

      odds [pair ]=observed/expected

  return odds
```

```python
sequence = "ATCGATCG"

ratio = odds_ratio(sequence)

print(ratio)
```
