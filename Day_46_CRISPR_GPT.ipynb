```python
# ===============================================
# CRISPR-GPT Repository Explorer
# Google Colab Demo
# ===============================================

import os
import subprocess
```

```python
REPO_NAME = "crispr-gpt-pub"
REPO_URL  = "https://github.com/cong-lab/crispr-gpt-pub.git"
# ================================================================
# Clone repository (only if not already cloned)
# ================================================================

if not os.path.exists(REPO_NAME):
    print("Cloning CRISPR_GPT respository....\n")
    subprocess.run(["git", "clone", REPO_URL])
else:
    print("Repository already exists.\n")
repo = REPO_NAME
# --------------------------------------------------------------
# Display Repository Structure
# -------------------------------------------------------------
print("="* 70)
print("                  CRISPR-GPT Repository Structure")
print("="* 70)
for root,dirs, files in os.walk(repo):
  level = root.replace(repo, '').count(os.sep)
  indent = ' '  * (level)
  print(f"{indent} {os.path.basename(root)}")
  subindent = ' ' * (level + 1)
  for file in files:
    print(f"{subindent} {file}")
```

```python
# ---------------------------------------------------------------------
# Explain Important Files
# --------------------------------------------------------------------
print("\n")
print("="*70)
print("Important Files")
print("="*70)

important ={
    "README.md":
      "Project documentation",
    "requirements.txt":
      "Python package dependencies",
    "main.py":
      "Application entry point (if available)",
    ".env":
      "Stores API Keys (created by user)",
    "llm.py":
      "LLM interface (if present)",
    "util.py":
      "Utility/helper functions"
}
for file, description in important.items():
  print(f"\n {file}")
  print(f" -> {description}")
```

```python
# ----------------------------------------------------------
# Count Files
# ---------------------------------------------------------

total_files =0
for _, _, files in os.walk(repo):
  total_files += len(files)
print("\n")
print("="*70)
print(f"\nTotal Files: {total_files}")
print("="*70)
print("\n Demo Completed Successfully!")
```

```python
"""
Demo 3: CRISPR-GPT Multi-Agent workflow Simulator
"""

import time
print("=" * 70)
print("                      CRISPR-GPT Multi-Agent workflow Simulator")
print("=" * 70)
task = input("\n Enter CRISPR Task:\n>")
print("\n")
print("="* 70)
print("Planner Agent")
print("="*70)

planner =[
    "Understand user objective",
    "Retrieve targe gene",
    "Identify editing strategy",
    "Locate target region",
    "Generate workflow"
]

for step in planner:
  print("Planning:",step)
  time.sleep(1)

print("\n")
print("="*70)
print("Execution Agent")
print("="*70)

executor =[
  "Retrieve sequence",
  "Search PAM Sites",
  "Generate guide RNAs",
  "Predict off-targets",
  "Rank candidate guides"
]
for step in executor:
  print("Executing:",step)
  time.sleep(1)
print("\n")
print("="*70)
print("Tool Provider")
print("="*70)

tools ={
    "Sequence Retrieval": "Ensembl",
    "Genome Browser": "UCSC",
    "Guide Design": "CHOPCHOP",
    "Off-target": "CRISPOR",
    "Alignment": "Bowtie2",
    "Visualization": "IGV"
}

for task_name, tool in tools.items():
  print(f"{task_name:25} --->{tool}")
  time.sleep(0.7)

print("\n")
print("=" *70)
print("Scientific Reporter")
print("="*70)
```
